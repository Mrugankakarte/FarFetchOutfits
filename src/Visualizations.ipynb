```python
import random

import matplotlib.image as mpimg
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
product_df = pd.read_parquet("data/graph/products.parquet")
product_df.head(2)
```

```python

def _plot_outfit(incomplete_outfit, missing_product, candidates):

    fig = plt.figure(figsize=(40, 12))
    gs = fig.add_gridspec(3, max(len(incomplete_outfit), len(candidates)))
    base_path = "../data/images/"
    for i in range(len(incomplete_outfit)):
        path = base_path + str(product_df.loc[incomplete_outfit[i]]['product_image_path'])
        image = mpimg.imread(path)
        ax = fig.add_subplot(gs[0, i])
        ax.imshow(image)
        ax.set_title(f"node id: {incomplete_outfit[i]}")

    path = base_path + str(product_df.loc[missing_product]['product_image_path'])
    image = mpimg.imread(path)
    ax = fig.add_subplot(gs[1,1])
    ax.imshow(image)
    ax.set_title(f"node id: {missing_product}")

    for j in range(len(candidates)):
        path = base_path + str(product_df.loc[candidates[j]]['product_image_path'])
        image = mpimg.imread(path)
        ax = fig.add_subplot(gs[2, j])
        ax.imshow(image)
        ax.set_title(f"node id: {candidates[j]}")


    plt.plot()
```

### Candidates from same category

```python
sample_valid = pd.read_parquet("ranked_samples.parquet")
sample_valid.head(2)
```

```python

idx = random.choice(np.arange(sample_valid.shape[0]))

_plot_outfit(sample_valid.loc[idx]['incomplete_outfits_nodeids'], sample_valid.loc[idx]['missing_prod_nodeid'], sample_valid.loc[idx]['ranked_candidates'])
```

### Candidates from all the catalog

```python
sample_valid2 = pd.read_parquet("ranked_samples_notsamecat.parquet")
sample_valid2.head(2)
```

```python

idx = random.choice(np.arange(sample_valid2.shape[0]))

_plot_outfit(sample_valid2.loc[idx]['incomplete_outfits_nodeids'], sample_valid2.loc[idx]['missing_prod_nodeid'], sample_valid2.loc[idx]['ranked_candidates'])
```
